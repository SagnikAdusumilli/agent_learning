```python
from dotenv import load_dotenv
_ = load_dotenv()
```

```python
from langgraph.graph import StateGraph, START, END

# Python built-in modules
from typing import TypedDict, Annotated
import operator

# LangChain messages
from langchain_core.messages import (
    AnyMessage,
    SystemMessage,
    HumanMessage,
    AIMessage,
    ToolMessage
)

# Ollama integration
from langchain_ollama import ChatOllama

# LangChain tools
from langchain_core.tools import tool
from langchain_tavily import TavilySearch
```

```python
search_tool = TavilySearch(max_results=5)
```

```python
@tool
def search_stocks(query: str) -> str:
    """Search the internet for company information and GICS sectors."""
    results = search_tool.invoke({"query": query})
    return str(results)
```

```python
class AgentState(TypedDict):
    messages: Annotated[list[AnyMessage], operator.add]
```

```python

class Agent:

    def __init__(self, model, tools, system=""):
        self.system = system

        # Create graph with AgentState schema
        graph = StateGraph(AgentState)

        # Register nodes
        graph.add_node("llm", self.call_ollama)
        graph.add_node("action", self.take_action)

        # Conditional routing after LLM
        graph.add_conditional_edges(
            "llm",
            self.exists_action,
            {True: "action", False: END}
        )

        # Return to LLM after executing a tool
        graph.add_edge("action", "llm")

        # Starting point
        graph.set_entry_point("llm")

        # Compile graph
        self.graph = graph.compile()

        # Register tools by name
        self.tools = {t.name: t for t in tools}

        # Bind tools to the model
        self.model = model.bind_tools(tools)

    def call_ollama(self, state: AgentState):
        messages = state["messages"]

        if self.system:
            messages = [SystemMessage(content=self.system)] + messages

        # Invoke LLM with conversation history
        message = self.model.invoke(messages)

        # LangGraph appends new messages using operator.add
        return {"messages": [message]}

    def exists_action(self, state: AgentState):
        result = state["messages"][-1]
        return len(result.tool_calls) > 0

    def take_action(self, state: AgentState):
        tool_calls = state["messages"][-1].tool_calls
        results = []

        for t in tool_calls:
            print(f"Calling: {t}")

            if t["name"] not in self.tools:
                print("\n ....bad tool name....")
                result = "bad tool name, retry"
            else:
                result = self.tools[t["name"]].invoke(t["args"])

            results.append(
                ToolMessage(
                    tool_call_id=t["id"],
                    name=t["name"],
                    content=str(result)
                )
            )

        print("Back to the model!")
        return {"messages": results}




            
```

```python
prompt = """
You are a stock sector classification assistant.

Your task is to classify stock ticker symbols into their
corresponding GICS sectors.

You have access to a web search tool.

For each ticker:
1. Use the search engine to identify the company and stock exchange.
2. Use the search engine to verify the company's GICS sector.
3. Assign exactly one of the 11 GICS sectors.
4. If the ticker is ambiguous, ask for clarification.
5. If the ticker is an ETF, identify it as an ETF rather
   than automatically assigning a single sector.
6. If uncertain, clearly say so rather than guessing.

Return results in this format:

Ticker | Company | Sector

Keep your response concise.
"""

prompt += """
IMPORTANT:
- Return exactly one row for every ticker requested.
- Never omit a ticker.
- Preserve the original order of the tickers.
- Before responding, check that the number of output rows
  equals the number of input tickers.
- If a sector cannot be verified, write "Unknown".
"""

model = ChatOllama(
    model="qwen3:8b",
    temperature=0,
    # no think.. just do lol
    reasoning=False
)

abot = Agent(model, [search_stocks], system=prompt)
```

```python
from IPython.display import Image

Image(abot.graph.get_graph().draw_png())
```

```python
result = abot.graph.invoke({
    "messages": [
        HumanMessage(content="Find the GICS sector for AAPL using web search.")
    ]
})

print(result["messages"][-1].content)
```

```python
from langchain_core.messages import HumanMessage

tickers = ["AAPL", "NVDA", "JPM", "XOM", "WMT"]

result = abot.graph.invoke({
    "messages": [
        HumanMessage(
            content=f"Use web search to verify the GICS sectors for these stocks: {', '.join(tickers)}. Classify each ticker."
        )
    ]
}, config={"recursion_limit": 60})

print(result["messages"][-1].content)
```
